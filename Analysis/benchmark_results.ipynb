```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# ==================================================
# LOAD
# ==================================================

df = pd.read_csv(
    "/oscar/data/epavlick/aparasel/Binding-Problem-Brown/Results/qwen_search_results.csv"
)

# ==================================================
# SUMMARY
# ==================================================

summary = (
    df.groupby(["condition", "num_distractors"])["correct"]
      .agg(["mean", "std", "count"])
      .reset_index()
)

summary["sem"] = (
    summary["std"] / np.sqrt(summary["count"])
)

# ==================================================
# PLOT
# ==================================================

plt.figure(figsize=(8, 5))

for condition in ["disjunctive", "conjunctive"]:

    subset = (
        summary[summary["condition"] == condition]
        .sort_values("num_distractors")
    )

    x = subset["num_distractors"].values
    y = subset["mean"].values
    sem = subset["sem"].values

    plt.plot(
        x,
        y,
        marker="o",
        linewidth=3,
        label=condition.capitalize(),
    )

    plt.fill_between(
        x,
        y - sem,
        y + sem,
        alpha=0.25,
    )
plt.axhline(
    y=0.5,
    color="black",
    linestyle="--",
    linewidth=2,
    alpha=0.8,
    label="Chance (50%)",
)
plt.xlabel("Number of Distractors")
plt.ylabel("Accuracy")
plt.title("Visual Search Accuracy")
plt.ylim(0, 1.05)
plt.grid(alpha=0.3)
plt.legend()

plt.tight_layout()
plt.show()
```

```python
import pandas as pd

df = pd.read_csv(
    "/oscar/data/epavlick/aparasel/Binding-Problem-Brown/Results/qwen_search_results_counterfactual_binary.csv"
)

# Keep only the fields we need
pairs = (
    df[
        [
            "condition",
            "trial",
            "target_present",
            "num_distractors",
            "correct",
        ]
    ]
    .pivot(
        index=["condition", "trial", "target_present"],
        columns="num_distractors",
        values="correct",
    )
    .reset_index()
)

# Rename columns for readability
pairs = pairs.rename(
    columns={
        10: "correct_10",
        40: "correct_40",
    }
)

# Cases where 40 succeeds but 10 fails
forty_better = pairs[
    (~pairs["correct_10"]) &
    (pairs["correct_40"])
]

print(f"40 correct / 10 incorrect: {len(forty_better)}")
display(forty_better)

# Also the opposite
ten_better = pairs[
    (pairs["correct_10"]) &
    (~pairs["correct_40"])
]

print(f"10 correct / 40 incorrect: {len(ten_better)}")
counts = (
    forty_better["target_present"]
    .value_counts()
    .rename(index={True: "GT=True", False: "GT=False"})
)

percents = (
    forty_better["target_present"]
    .value_counts(normalize=True)
    .rename(index={True: "GT=True", False: "GT=False"})
    * 100
)

summary = pd.DataFrame({
    "Count": counts,
    "Percent": percents.round(1),
})

display(summary)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# ==================================================
# LOAD
# ==================================================

df = pd.read_csv(
    "/oscar/data/epavlick/aparasel/Binding-Problem-Brown/Results/qwen_search_results.csv"
)

# ==================================================
# SUMMARY
# ==================================================

summary = (
    df.groupby(
        ["target_present", "condition", "num_distractors"]
    )["correct"]
    .agg(["mean", "std", "count"])
    .reset_index()
)

summary["sem"] = (
    summary["std"] / np.sqrt(summary["count"])
)

# ==================================================
# PLOT
# ==================================================

fig, axes = plt.subplots(
    1,
    2,
    figsize=(12, 5),
    sharey=True,
)

for ax, target_present in zip(
    axes,
    [True, False]
):

    panel = summary[
        summary["target_present"] == target_present
    ]

    for condition in ["disjunctive", "conjunctive"]:

        subset = (
            panel[
                panel["condition"] == condition
            ]
            .sort_values("num_distractors")
        )

        x = subset["num_distractors"].values
        y = subset["mean"].values
        sem = subset["sem"].values

        ax.plot(
            x,
            y,
            marker="o",
            linewidth=3,
            label=condition.capitalize(),
        )

        ax.fill_between(
            x,
            y - sem,
            y + sem,
            alpha=0.25,
        )

    ax.axhline(
        y=0.5,
        color="black",
        linestyle="--",
        linewidth=2,
        alpha=0.8,
    )

    title = (
        "Target Present"
        if target_present
        else "Target Absent"
    )

    ax.set_title(title)
    ax.set_xlabel("Number of Distractors")
    ax.grid(alpha=0.3)
    ax.set_ylim(0, 1.05)

axes[0].set_ylabel("Accuracy")
axes[1].legend()

plt.suptitle(
    "Visual Search Accuracy Conditioned on Target Presence",
    y=1.02,
)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# ==================================================
# LOAD
# ==================================================

df = pd.read_csv(
    "/oscar/data/epavlick/aparasel/Binding-Problem-Brown/Results/qwen_last_input_probe.csv"
)

# ==================================================
# CONFIG
# ==================================================

distractor_counts = sorted(df["num_distractors"].unique())

fig, axes = plt.subplots(
    2,
    5,
    figsize=(18, 7),
    sharex=True,
    sharey=True,
)

axes = axes.flatten()

# ==================================================
# PLOT
# ==================================================

for ax, nd in zip(axes, distractor_counts):

    subset = df[
        df["num_distractors"] == nd
    ]

    for condition in [
        "disjunctive",
        "conjunctive",
    ]:

        curve = (
            subset[
                subset["condition"] == condition
            ]
            .sort_values("layer")
        )

        ax.plot(
            curve["layer"],
            curve["test_bal_acc"],
            marker="o",
            linewidth=2.5,
            label=condition.capitalize(),
        )

    ax.axhline(
        0.5,
        color="black",
        linestyle="--",
        linewidth=1,
        alpha=0.6,
    )

    ax.set_title(
        f"{nd} Distractors",
        fontsize=12,
    )

    ax.set_ylim(0.45, 1.02)
    ax.grid(alpha=0.3)

# ==================================================
# LABELS
# ==================================================

fig.supxlabel(
    "Layer",
    fontsize=16,
)

fig.supylabel(
    "Balanced Probe Accuracy",
    fontsize=16,
)

handles, labels = axes[0].get_legend_handles_labels()

fig.legend(
    handles,
    labels,
    loc="upper center",
    ncol=2,
    frameon=False,
    fontsize=12,
)

plt.tight_layout(rect=[0, 0, 1, 0.93])

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# ==================================================
# LOAD
# ==================================================

df = pd.read_csv(
    "/oscar/data/epavlick/aparasel/Binding-Problem-Brown/Results/continuation_probe.csv"
)

# ==================================================
# CONFIG
# ==================================================

distractor_counts = sorted(
    df["num_distractors"].unique()
)

fig, axes = plt.subplots(
    2,
    5,
    figsize=(18, 7),
    sharex=True,
    sharey=True,
)

axes = axes.flatten()

# ==================================================
# PLOT
# ==================================================

for ax, nd in zip(axes, distractor_counts):

    subset = df[
        df["num_distractors"] == nd
    ]

    for condition in [
        "disjunctive",
        "conjunctive",
    ]:

        curve = (
            subset[
                subset["condition"] == condition
            ]
            .sort_values("layer")
        )

        ax.plot(
            curve["layer"],
            curve["test_bal_acc"],
            marker="o",
            linewidth=2.5,
            label=condition.capitalize(),
        )

        # ------------------------------------------
        # chance line for this condition
        # ------------------------------------------

        n_classes = (
            curve["num_classes"]
            .iloc[0]
        )

        chance = 1.0 / n_classes

        linestyle = (
            "--"
            if condition == "disjunctive"
            else ":"
        )

        ax.axhline(
            chance,
            linestyle=linestyle,
            linewidth=1.5,
            alpha=0.7,
        )

    ax.set_title(
        f"{nd} Distractors",
        fontsize=12,
    )

    ax.grid(alpha=0.3)

# ==================================================
# GLOBAL Y LIMITS
# ==================================================

ymin = (
    df["test_bal_acc"].min() - 0.05
)
ymax = (
    df["test_bal_acc"].max() + 0.05
)

for ax in axes:
    ax.set_ylim(
        max(0, ymin),
        min(1.02, ymax),
    )

# ==================================================
# LABELS
# ==================================================

fig.supxlabel(
    "Layer",
    fontsize=16,
)

fig.supylabel(
    "Balanced Probe Accuracy",
    fontsize=16,
)

handles, labels = (
    axes[0]
    .get_legend_handles_labels()
)

fig.legend(
    handles,
    labels,
    loc="upper center",
    ncol=2,
    frameon=False,
    fontsize=12,
)

plt.tight_layout(
    rect=[0, 0, 1, 0.93]
)

plt.show()
```

```python
# NO PADING 
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# ==================================================
# LOAD
# ==================================================

vision_df = pd.read_csv(
    "/oscar/data/epavlick/aparasel/Binding-Problem-Brown/Results/qwen_vlm_text_search_no_padding.csv"
)

language_df = pd.read_csv(
    "/oscar/data/epavlick/aparasel/Binding-Problem-Brown/Results/qwen_llm_text_search_no_padding.csv"
)

math_df = pd.read_csv(
    "/oscar/data/epavlick/aparasel/Binding-Problem-Brown/Results/qwen_math_text_search_no_padding.csv"
)

import re
import numpy as np

def extract_answer(text):
    if pd.isna(text):
        return np.nan

    text = str(text)

    # [True] or [False]
    match = re.search(
        r"\[\s*(True|False)\s*\]",
        text,
        re.IGNORECASE,
    )
    if match:
        return match.group(1).lower() == "true"

    # \boxed{\text{True}} or \boxed{\text{False}}
    match = re.search(
        r"\\boxed\{\s*\\text\{\s*(True|False)\s*\}\s*\}",
        text,
        re.IGNORECASE,
    )
    if match:
        return match.group(1).lower() == "true"

    # \boxed{True} or \boxed{False}
    match = re.search(
        r"\\boxed\{\s*(True|False)\s*\}",
        text,
        re.IGNORECASE,
    )
    if match:
        return match.group(1).lower() == "true"

    return np.nan


# recompute predictions
math_df["prediction"] = (
    math_df["response"]
    .apply(extract_answer)
)

# recompute correctness
math_df["correct"] = np.where(
    math_df["prediction"].isna(),
    np.nan,
    (
        math_df["prediction"]
        == math_df["target_present"]
    ).astype(float)
)

# make NaN predictions have NaN correctness
math_df.loc[
    math_df["prediction"].isna(),
    "correct"
] = np.nan

# ==================================================
# SUMMARY
# ==================================================

def summarize(df):
    summary = (
        df.groupby(["condition", "num_distractors"])["correct"]
          .agg(["mean", "std", "count"])
          .reset_index()
    )

    summary["sem"] = (
        summary["std"] / np.sqrt(summary["count"])
    )

    return summary

vision_summary = summarize(vision_df)
language_summary = summarize(language_df)
math_summary = summarize(math_df)

# ==================================================
# STYLE
# ==================================================

modality_info = {
    "VLM": {
        "summary": vision_summary,
        "color": "C0",
    },
    "LLM": {
        "summary": language_summary,
        "color": "C1",
    },
    "Math LLM": {
        "summary": math_summary,
        "color": "C2",
    },
}

linestyles = {
    "disjunctive": "-",
    "conjunctive": "--",
}

# ==================================================
# PLOT
# ==================================================

plt.figure(figsize=(8, 5))

for modality, info in modality_info.items():

    summary = info["summary"]
    color = info["color"]

    for condition in ["disjunctive", "conjunctive"]:

        subset = (
            summary[
                summary["condition"] == condition
            ]
            .sort_values("num_distractors")
        )

        x = subset["num_distractors"].values
        y = subset["mean"].values
        sem = subset["sem"].values

        plt.plot(
            x,
            y,
            color=color,
            linestyle=linestyles[condition],
            marker="o",
            linewidth=3,
            label=f"{modality} - {condition.capitalize()}",
        )

        plt.fill_between(
            x,
            y - sem,
            y + sem,
            color=color,
            alpha=0.15,
        )

# ==================================================
# CHANCE
# ==================================================

plt.axhline(
    0.5,
    color="black",
    linestyle=":",
    linewidth=2,
    label="Chance (50%)",
)

# ==================================================
# FORMATTING
# ==================================================

plt.xlabel("Number of Distractors")
plt.ylabel("Accuracy")
plt.title("Language Search: VLM vs LLM vs LLM-Math")
plt.ylim(0, 1.05)
plt.grid(alpha=0.3)
plt.legend()

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D

# ==================================================
# LOAD
# ==================================================

vision_df = pd.read_csv(
    "/oscar/data/epavlick/aparasel/Binding-Problem-Brown/Results/qwen_text_search_period_vlm.csv"
)

language_df = pd.read_csv(
    "/oscar/data/epavlick/aparasel/Binding-Problem-Brown/Results/qwen_text_search_period_lang.csv"
)

# ==================================================
# REMOVE TRIALS WITH NO PREDICTED LABEL
# ==================================================

def remove_no_label(df):

    df = df.copy()

    mask = (
        df["prediction"].notna()
        & (df["prediction"].astype(str).str.strip() != "")
    )

    print(f"Keeping {mask.sum()} / {len(df)} trials")

    return df.loc[mask]


vision_df = remove_no_label(vision_df)
language_df = remove_no_label(language_df)

# ==================================================
# SUMMARY
# ==================================================

def summarize(df):
    summary = (
        df.groupby(["condition", "num_distractors"])["correct"]
          .agg(["mean", "std", "count"])
          .reset_index()
    )

    summary["sem"] = (
        summary["std"] / np.sqrt(summary["count"])
    )

    return summary


vision_summary = summarize(vision_df)
language_summary = summarize(language_df)

# ==================================================
# PLOT
# ==================================================

plt.figure(figsize=(8, 5))

# Color encodes model type
colors = {
    "VLM": "C0",
    "LLM": "C1",
}

# Line style encodes search type
linestyles = {
    "disjunctive": "-",
    "conjunctive": "--",
}

datasets = [
    ("VLM", vision_summary),
    ("LLM", language_summary),
]

for model_name, summary_df in datasets:

    color = colors[model_name]

    for condition in ["disjunctive", "conjunctive"]:

        subset = (
            summary_df[
                summary_df["condition"] == condition
            ]
            .sort_values("num_distractors")
        )

        x = subset["num_distractors"].values
        y = subset["mean"].values
        sem = subset["sem"].values

        plt.plot(
            x,
            y,
            color=color,
            linestyle=linestyles[condition],
            marker="o",
            linewidth=3,
        )

        plt.fill_between(
            x,
            y - sem,
            y + sem,
            color=color,
            alpha=0.15,
        )

# ==================================================
# CHANCE
# ==================================================

plt.axhline(
    0.5,
    color="black",
    linestyle=":",
    linewidth=2,
)

# ==================================================
# CUSTOM LEGEND
# ==================================================

legend_elements = [
    Line2D(
        [0],
        [0],
        color=colors["VLM"],
        lw=3,
        label="VLM",
    ),
    Line2D(
        [0],
        [0],
        color=colors["LLM"],
        lw=3,
        label="LLM",
    ),
    Line2D(
        [0],
        [0],
        color="black",
        lw=3,
        linestyle="-",
        label="Disjunctive",
    ),
    Line2D(
        [0],
        [0],
        color="black",
        lw=3,
        linestyle="--",
        label="Conjunctive",
    ),
    Line2D(
        [0],
        [0],
        color="black",
        lw=2,
        linestyle=":",
        label="Chance (50%)",
    ),
]

plt.legend(
    handles=legend_elements,
    frameon=True,
)

# ==================================================
# FORMATTING
# ==================================================

plt.xlabel("Number of Distractors", fontsize=12)
plt.ylabel("Accuracy", fontsize=12)
plt.title(
    "Qwen Visual vs Language Search\n(Excluding No-Label Trials)",
    fontsize=14,
)
plt.ylim(0, 1.05)
plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
import torch
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.metrics.pairwise import (
    cosine_similarity,
    euclidean_distances,
)

# ==================================================
# LOAD
# ==================================================

records = torch.load(
    "/oscar/data/epavlick/aparasel/Binding-Problem-Brown/Results/qwen_target_reps.pt"
)

df = pd.DataFrame(records)

# ==================================================
# SETTINGS
# ==================================================

LAYER = 10

DISTRACTOR_COUNTS = [
    5, 10, 15, 20, 25,
    30, 35, 40, 45, 50,
]

# ==================================================
# FILTER
# ==================================================

subset = df[
    (df["condition"] == "conjunctive")
    &
    (df["layer"] == LAYER)
].copy()

# ==================================================
# AVERAGE TARGET REPRESENTATION
# ==================================================

means = []

for nd in DISTRACTOR_COUNTS:

    reps = subset[
        subset["num_distractors"] == nd
    ]["target_rep"].tolist()

    reps = torch.stack(
        reps
    ).float()

    mean_rep = (
        reps.mean(dim=0)
        .numpy()
    )

    means.append(
        mean_rep
    )

means = np.stack(
    means
)

# ==================================================
# METRICS
# ==================================================

cos_sim = cosine_similarity(
    means
)

l2_dist = euclidean_distances(
    means
)

means_norm = (
    means
    / np.linalg.norm(
        means,
        axis=1,
        keepdims=True,
    )
)

norm_l2_dist = euclidean_distances(
    means_norm
)

# ==================================================
# PLOT
# ==================================================

fig, axes = plt.subplots(
    1,
    3,
    figsize=(18, 5),
)

plots = [
    (
        cos_sim,
        "Cosine Similarity",
        "Cosine Similarity",
        (cos_sim.min(), 1.0),
    ),
    (
        l2_dist,
        "Raw L2 Distance",
        "Euclidean Distance",
        None,
    ),
    (
        norm_l2_dist,
        "Normalized L2 Distance",
        "Euclidean Distance",
        None,
    ),
]

for ax, (
    matrix,
    title,
    cbar_label,
    clim,
) in zip(
    axes,
    plots,
):

    if clim is None:
        im = ax.imshow(
            matrix,
            cmap="viridis",
        )
    else:
        im = ax.imshow(
            matrix,
            cmap="viridis",
            vmin=clim[0],
            vmax=clim[1],
        )

    ax.set_xticks(
        range(len(DISTRACTOR_COUNTS))
    )

    ax.set_yticks(
        range(len(DISTRACTOR_COUNTS))
    )

    ax.set_xticklabels(
        DISTRACTOR_COUNTS
    )

    ax.set_yticklabels(
        DISTRACTOR_COUNTS
    )

    ax.set_xlabel(
        "# Distractors"
    )

    ax.set_ylabel(
        "# Distractors"
    )

    ax.set_title(
        f"Layer {LAYER}\n{title}"
    )

    fig.colorbar(
        im,
        ax=ax,
        label=cbar_label,
    )

plt.tight_layout()
plt.show()
```

```python
import torch
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.metrics.pairwise import (
    cosine_similarity,
    euclidean_distances,
)

# ==================================================
# LOAD
# ==================================================

records = torch.load(
    "/oscar/data/epavlick/aparasel/Binding-Problem-Brown/Results/qwen_target_reps.pt"
)

df = pd.DataFrame(records)

# ==================================================
# SETTINGS
# ==================================================

layers = sorted(
    df["layer"].unique()
)

cos_5_50 = []
l2_5_50 = []
norm_l2_5_50 = []

# ==================================================
# LOOP OVER LAYERS
# ==================================================

for layer in layers:

    subset = df[
        (df["condition"] == "conjunctive")
        &
        (df["layer"] == layer)
    ]

    reps_5 = torch.stack(
        subset[
            subset["num_distractors"] == 5
        ]["target_rep"].tolist()
    ).float()

    reps_50 = torch.stack(
        subset[
            subset["num_distractors"] == 50
        ]["target_rep"].tolist()
    ).float()

    mean_5 = (
        reps_5.mean(dim=0)
        .numpy()
    )

    mean_50 = (
        reps_50.mean(dim=0)
        .numpy()
    )

    # ------------------------------------------
    # cosine
    # ------------------------------------------

    cos = cosine_similarity(
        mean_5[None],
        mean_50[None],
    )[0, 0]

    cos_5_50.append(
        cos
    )

    # ------------------------------------------
    # raw L2
    # ------------------------------------------

    l2 = np.linalg.norm(
        mean_5 - mean_50
    )

    l2_5_50.append(
        l2
    )

    # ------------------------------------------
    # normalized L2
    # ------------------------------------------

    mean_5_norm = (
        mean_5
        / np.linalg.norm(mean_5)
    )

    mean_50_norm = (
        mean_50
        / np.linalg.norm(mean_50)
    )

    norm_l2 = np.linalg.norm(
        mean_5_norm
        -
        mean_50_norm
    )

    norm_l2_5_50.append(
        norm_l2
    )

# ==================================================
# PLOT
# ==================================================

fig, axes = plt.subplots(
    1,
    3,
    figsize=(16, 4)
)

axes[0].plot(
    layers,
    cos_5_50,
    marker="o",
    linewidth=3,
)

axes[0].set_title(
    "Cosine Similarity\n5 vs 50 Distractors"
)

axes[0].set_xlabel(
    "Layer"
)

axes[0].set_ylabel(
    "Cosine Similarity"
)

axes[0].grid(alpha=0.3)

# ------------------------------------------------

axes[1].plot(
    layers,
    l2_5_50,
    marker="o",
    linewidth=3,
)

axes[1].set_title(
    "Raw L2 Distance\n5 vs 50 Distractors"
)

axes[1].set_xlabel(
    "Layer"
)

axes[1].set_ylabel(
    "L2 Distance"
)

axes[1].grid(alpha=0.3)

# ------------------------------------------------

axes[2].plot(
    layers,
    norm_l2_5_50,
    marker="o",
    linewidth=3,
)

axes[2].set_title(
    "Normalized L2 Distance\n5 vs 50 Distractors"
)

axes[2].set_xlabel(
    "Layer"
)

axes[2].set_ylabel(
    "Normalized L2"
)

axes[2].grid(alpha=0.3)

plt.tight_layout()
plt.show()
```
